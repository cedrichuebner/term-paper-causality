# Does optional individual taxation increase female labor supply?

Evidence from the 2018 Luxembourg Reform

Cédric Hübner  
Nadine Theisen  
Kwangho Yo  
August 31, 2026

Increasing women’s participation in the labor market is an important policy goal. However, in countries where married couples are taxed based on their shared income instead of individually, the nature of that system can disincentivize women from working more. Previous reforms toward individual taxation have proven effective in increasing the working hours of women. This paper investigates whether a 2018 reform in Luxembourg, which gave married couples the option to choose between joint and individual taxation, had the same effect. Based on models of utility maximization and intra-household bargaining, we assume that such a reform leads to an increase in working hours even though the change is not mandatory. Our analysis uses a difference-in-differences approach on individual-level data from Luxembourg and five other countries between 2012 and 2022. The resulting models show no effect of the reform on the working hours of women when controlling for education and children. However, the results are inconclusive, primarily due to methodological limitations and potential external influences.

## 1 Introduction

Taxes are not only the most important source of state revenue, but their design also influences individuals’ behavior, either intentionally or accidentally. Income taxes are no exception. In simplified terms, individuals calculate whether the additional income they would earn by working more hours is worth the extra time spent compared to how they could have spent that time otherwise. As additional income is influenced by tax rates, the design of the tax system affects individual labor supply, i.e., how much paid work a person does per week. Therefore, increasing the individual labor supply becomes an important policy goal for states, not only to increase tax revenue, but also to address the growing issues caused by demographic changes, such as a shortage of skilled workers and mounting pressure on social security systems.

Women have the greatest untapped potential for higher labor force activity. One way to increase women’s working hours is through the tax system. In some countries such as Germany or France, the rate of taxation for married couples is calculated based on their combined income. This creates a disincentive for secondary earners, who are usually women, to work more hours, which often leads to them taking on more domestic work and childcare. In addition to negatively impacting working hours, the design of the tax system perpetuates gender inequality, by causing lower individual income, pension payments, and consequently, less financial security for women later in life.

To address these suboptimal outcomes and achieve higher workforce activity, critics often advocate shifting from joint to individual taxation. The most recent example of a country following this direction is Luxembourg: since 2018, married couples have the possibility to voluntarily choose to be taxed individually. This paper examines how the reform has affected female labor activity, investigating whether shifting toward optional individual taxation is sufficient to increase women’s working hours.

The first part of the paper introduces the theoretical argument. Based on the concept of individual utility maximization and intra-household bargaining, we hypothesize that women’s working hours will increase even under optional individual taxation. The next section is devoted to the research design. Here we present the difference-in-differences design used to analyze the effects of the tax reform, argue why it is suited to determine causal effects, and lay out underlying assumptions. In addition, we introduce the data and variables used for the analysis. The following results and robustness checks show that there is no reliable evidence that introducing the option for individual taxation significantly increased female working hours as a result of the 2018 reform. We conclude by discussing and summarizing our findings.

## 2 The 2018 Luxembourg Income Tax Reform

Luxembourg is a particularly relevant case regarding its evolving tax system. Historically, married couples were taxed jointly through compulsory income splitting, while registered partners could also opt for joint taxation under certain conditions. Under joint taxation with income splitting, both spouses’ incomes are added together and divided by two. The progressive tax rate is then applied to this half, and the resulting tax amount is doubled to determine the couple’s total tax liability. As the income tax scale is progressive, the benefit of splitting income depends on the difference between the spouses’ incomes. This means conversely that, for couples with equal incomes, the tax rate applied to half of their total taxable income is the same. Where there are significant differences in income, this results in an advantage at the couple level, as the main earner’s income is placed in a lower tax bracket because it is treated as if both partners had each earned half of the total taxable income. However, the amount of the benefit also depends on total taxable income and the structure of the progressive tax scale, meaning that the effect is not linear.

Since the 2018 tax year, married couples have been able to choose among joint taxation with income splitting, pure individual taxation, and income reallocation ([El Mashlohi et al., 2026](#ref-euromod2022)). With individual taxation, each spouse files their income tax separately. Under individual taxation with reallocation, the couple’s total adjusted taxable income is first determined jointly and then allocated between the spouses. Each spouse is subsequently taxed separately on their allocated share. If no other allocation is specified, the income is divided equally.

A further step toward individual taxation is planned for 2028 to replace the existing tax classes with a single tax class and individual tax, starting with newly married and registered couples. Taxpayers who were jointly taxed before 2028 would be allowed to remain under the previous tariff for a transition period of 25 years, assuming they do not switch permanently to the individual assessment earlier ([Ministère des Finances, 2026](#ref-ministeredesfinances2026)).

Existing research also suggests that full individualization could have meaningful effects on labor supply in Luxembourg. Islam et al. ([2020](#ref-islam2020)) simulates a reform from joint to individual taxation and estimates an increase of around 2.27% in married women’s labor supply, while married men’s labor supply remains largely unchanged. This way, the case could be insightful since the 2018 reform introduced individual taxation as an option on top of the marital splitting. The optional model is less explored in existing literature on replacing reforms (e.g., [Kalíšková, 2014](#ref-kaliskova2014); [Crossley & Jeon, 2007](#ref-crossley2007); [Selin, 2014](#ref-selin2014)). So, the 2018 reform in Luxembourg provides a case to test whether optional economic individualization, meaning individual independence over household utility here, suffices to generate measurable labor-supply effects.

## 3 Models of utility maximization and the distribution of labor under different tax systems

A country’s specific tax system has a great influence on individual behavior. Individuals make decisions in the allocation of resources, maximizing utility given income and prices. When conditions change, choices are reoriented toward a new best set. In this logic, one’s behavior reflects a set of income and prices, pursuing a utility-maximizing option ([Samuelson, 1938, p. 6](#ref-samuelson1938)). Because income is a key determinant of consumption, and consumption is a core part of utility ([Hausman, 1985, p. 4](#ref-hausman1985)), any change in the tax system that alters wages reshapes the individual utility. Thus, the tax system fundamentally affects the utility assessment process. In this light, tax models are central to understanding how individuals respond to policy change.

From this economic point of departure and with the background on how joint taxation affects the household and spouses individually, the next sections elaborate further on the theoretical assumptions from the body of literature that build the foundation for our argument. In what follows, our hypotheses are presented along with the theoretical argument.

### 3.1 Female labor activity under individual taxation

The nature of the tax system plays a significant role in individual decision-making processes, but effects oftentimes vary between genders. One of the better explored topics is the change in female labor supply with the reform of personal income tax for married couples. Previous research asserts that the general tendency of female labor supply, especially among married women, is assumed to rise with the introduction of individual taxation ([Bick & Fuchs-Schündeln, 2017](#ref-bick2017); [LaLumia, 2008, p. 11](#ref-lalumia2008); [Selin, 2014, p. 22](#ref-selin2014)), whereas male labor supply does not seem to be affected ([Kalíšková, 2014, p. 2](#ref-kaliskova2014); [LaLumia, 2008, p. 11](#ref-lalumia2008)). In theoretical accounts, an increase in female labor supply is unambiguous following a change in tax rate.

When the tax system changes, the supply of labor is mainly determined by two effects: the substitution and income effects. In calculations on the allocation of labor and leisure, the opportunity cost of choosing one over the other is considered. Here, on the extensive margin, the female labor supply is primarily driven by the substitution effect, since the income effect does not exist because of non-participation in the labor market. Moreover, the elasticity of wages matters in magnitude. Women are conventionally assumed to have a comparative advantage in domestic labor over men, so that their labor supply is more elastic to income changes than men’s ([Alesina et al., 2011, p. 4](#ref-alesina2011)), consistent with empirical findings ([Kalíšková, 2014](#ref-kaliskova2014); [LaLumia, 2008](#ref-lalumia2008)).

### 3.2 Effects of household decision-making on female labor

Although individual taxation is expected to increase the number of hours women work, it is unclear what happens when both options exist within a tax system. The choices of married couples can be explained by looking at economic models of household behavior. The model of joint taxation with income splitting complements a traditional unitary household model in which spouses specialize in different forms of work, with one partner primarily in paid employment and the other more engaged in unpaid care and domestic work. Becker’s ([1974](#ref-becker1974)) household model explains the specialization through efficiency gains from the division of labor. In contrast, Iversen and Rosenbluth’s ([2006](#ref-iversen2006)) bargaining model is oriented at the intra-household perspective with the individual as the relevant unit ([2006, pp. 1–2](#ref-iversen2006)). Based on this premise, joint taxation can be conceptualized as a policy that incentivizes couples to optimize their total disposable income. This policy maximizes total household disposable income but does not necessarily improve the individual economic position of spouses. However, the traditional household model does not align with historical and theoretical shifts in the gender gap in labor market activity, economic dependence within marriage, and longevity of marriages in western welfare states.

In their bargaining model, Iversen & Rosenbluth ([2006, pp. 1–4](#ref-iversen2006)) conceptualize that if marriage is a contract that can be dissolved, the assumption of unitary interests between spouses falls short. Instead of a couple with one shared set of interests, the two partners are conceptualized as individuals with their own economic interests, leading to a bargaining understanding of the distribution of work. Applied to income taxation, the secondary earner would be expected to prefer individual taxation if it maximizes individual economic assets even if the benefit in household income would be larger if taxed jointly.

At the macro level, Edlund & Pande ([2002](#ref-edlund2002)) argues that while marriage historically pooled economic interests, the decline of dependence can make individual interests more salient, contributing to women supporting redistribution more than before compared to men. Individual taxation is more likely to be opted for if the perspective of the secondary earner shapes the decision and the unitary model loses importance.

In addition to considerations related to maximizing disposable income, two more factors determine the level of female labor activity. First, higher levels of education tend to be associated with higher pay. As this increases the utility of paid work, individuals are more likely to shift their priorities away from unpaid work in the household ([Busso & Romero Fonseca, 2015, p. 204](#ref-busso2015)).

The second factor impacting women’s working hours is family planning. Having children not only leads mothers (and to some extent also fathers) to leave the workforce temporarily around the time of birth. It also influences labor supply since it incurs the opportunity cost of working outside the home. Furthermore, in a context of different relative wage returns of partners in the household, the one with the lower income prospects is more often to go out of work to some extent ([Busso & Romero Fonseca, 2015, p. 204](#ref-busso2015)).

### 3.3 Hypotheses

Taken together, this paper aims to test the impact of the introduction of optional individual taxation on female labor supply in Luxembourg. Therein, it differs from previous studies, which examine changes to compulsory individual filing. Since the Luxembourg case allows couples to keep filing jointly, it is worth exploring whether the outcomes are captured by the main theoretical discourses.

Based on optimal utility theory in taxation, it is assumed that women tend to work more under individual taxation. First, under optional individual taxation, individuals may project their potential earnings according to Nash-Pareto. Second, there is a possibility that the income of the secondary earner would fall into a higher tax bracket under joint taxation, which undermines Pareto-optimality. Therefore, it is expected that women are more likely to select individual taxation, increasing labor supply.

Drawing on the bargaining model, it is anticipated that women are inclined to choose individual taxation. The traditional assumption of family as an economic unit has diminished with increasing female economic independence. In other words, a unitary family is no longer the minimum unit; instead, intra-household bargaining enables individuals to prioritize and maximize their utility and individual outside options. The relative position of the secondary earner in the bargain might open the opportunity for decision-making that prioritizes individual income, even if the household net income decreases. To sum up, we hypothesize:

H1: Married women’s weekly working hours will increase under optional individual taxation.

H2: Married men’s working hours will remain constant with the introduction of optional individual taxation.

## 4 Research Design

To examine the effects of the Luxembourg tax reform on working hours, we estimate a difference-in-differences model (below DiD). This approach is commonly used to analyze causal effects in situations where random assignment into treatment and control groups is not feasible. The method has therefore been widely applied in policy change research in general, as well as in research examining the impact of tax reforms on labor supply in particular (see for example [Herold et al., 2025](#ref-herold2025); [Crossley & Jeon, 2007](#ref-crossley2007); [Doorley, 2018](#ref-doorley2018); [Kalíšková, 2014](#ref-kaliskova2014); [Selin, 2014](#ref-selin2014)).

In a DiD design, there is a treatment group and a control group, and the outcome variable is observed for both before and after the treatment. In our case, the treatment is the 2018 tax reform with the introduction of optional individual taxation and the outcome is female working hours. The treatment and control groups can vary in their initial level of the outcome. After the policy change, the treatment group is expected to deviate from the control group if the reform has an effect, meaning that the difference in the outcome variable is smaller or larger than before. In our case, the DiD design is able to capture whether female employment in Luxembourg increased or decreased more strongly after the reform than it did in comparable untreated countries, while allowing for pre-existing differences in employment levels between countries.

We deem this methodological approach to be appropriate for two reasons. First, it removes time-invariant differences between sample groups. Since our case is based on non-random assignment, it is subject to time-varying confounders: for example, a seemingly irrelevant change might have a significant impact on female labor. This, in turn, leads to a high possibility of misattribution in estimation. Second, the method allows us to obtain counterfactual estimates. By adopting the parallel trends assumption between treatment and control groups, it projects hypothetical estimates from the treated group where no treatment exists. By doing so, it enables us to compare actual trends and counterfactual estimates in our case. In this regard, DiD appears to be the most suitable approach to test the impact of tax reform in Luxembourg.

### 4.1 Data and operationalization

For our analysis, we rely on individual-level data from the Luxembourg Income Study, which aggregates and harmonizes a number of national cross-sectional surveys into a single dataset. We use data from Luxembourg for the treatment group and data from Belgium, France, Germany, Spain, and Ireland for the control group. Both were collected repeatedly between 2012 and 2022. Because each survey is based on a randomly taken sample from the same population, the “earlier cross-section can be used as a surrogate for the individuals in the treatment and control groups in the later cross-section” ([Stock, 2020, p. 494](#ref-stock2020)). Including multiple countries in the control group has the advantage of delivering more precise estimates.

The specific model we use for our analysis is a two-way fixed effects model of the following form: $y_{it} = \beta_1 x_{it} + \alpha_i + \delta_t + X'_{it} + u_{it}$

The data are operationalized as follows: As the main outcome $y$ we use a variable for working hours. Here, respondents are asked how many hours per week they work in their main job.

As for the independent variable, we first create a binary dummy variable $x_{it}$ that captures the treatment status. Here, all observations from married individuals in Luxembourg who lie in the post-treatment period, i.e., after the tax reform, are coded with the value one. The remaining observations from Luxembourg in the pre-treatment period and the control countries in both periods are coded as zero, as neither of those experienced the tax reform. In our models, the coefficient of this treatment variable indicates the effect of the reform on the overall population, rather than on individuals who actually opt for individual taxation. The coefficients therefore represent an intent-to-treat effect that captures the effect of having the option to file taxes individually, including couples who continued to file jointly. Thus, incomplete take-up might underestimate the effect of couples who actually opt for individual filing.

For the country-fixed effects $\alpha_i$ as well as the time-fixed effects $\delta_t$, we rely on variables in the dataset, capturing the individual’s specific country and the reference year of the respective survey. As controls $X'_{it}$, we use individual-level variables capturing levels of education and a binary variable indicating whether children are present in the household. Including these controls not only addresses potential confounding effects but also has the function of minimizing potential distortions caused by differences in the distribution of these variables over time and between countries ([Bosco & Maranzano, 2025](#ref-bosco2025)). In order to approximate the results to the overall population, we used the normalized, individual-level weight variable provided by the dataset.

### 4.2 Methodological assumptions

The DiD method depends on a number of assumptions that need to be fulfilled to obtain reliable results: First, the parallel trend assumption supposes that the outcome of interest would have taken the same trajectory in both the treatment and control group in the absence of a treatment ([Callaway, 2022, p. 3](#ref-callaway2022)). In our case, this means that the average working hours would have developed at the same rate in Luxembourg as in the control countries if the reform had not been implemented. Since the assumption is based on a counterfactual, it cannot be empirically tested for designs that rely on the comparison of two time periods. However, a graphical comparison of the pre-treatment trend can be used as a rough approximation to show parallel trends before the treatment ([Wing et al., 2018, p. 459](#ref-wing2018)). We present such a comparison as part of our robustness checks in the result section.

A second assumption is that individuals in the treatment group are unable to anticipate the treatment ([Bosco & Maranzano, 2025, p. 6](#ref-bosco2025)). In theory, individuals could adjust their weekly working hours between the announcement of the policy and its implementation. However, we expect that individuals cannot flexibly adjust the hours they work in a short period of time, so anticipation effects are likely to be negligible.

Further, it is assumed that selection into treatment has to be exogenous and not made by the individuals themselves. Since marital status is a criterion for joint taxation and the reform, individuals can in principle self-select into or out of treatment through marriage or divorce. The reform might have altered couples’ incentives to get married. For example, couples with a preference for individual taxation might have avoided getting married before the reform. We therefore cannot fully rule out selection bias arising from endogenous marital status.

Lastly, the DiD method is based on the assumption that there are no spillover effects, meaning that the treatment only affects individuals in the treatment group but not in the control group ([Bosco & Maranzano, 2025, p. 15](#ref-bosco2025)). This is potentially problematic as a sizable share of Luxembourg’s working population consists of cross-border commuters ([Leduc & Sierminska, 2025, p. 1](#ref-leduc2025)). Therefore, individuals who, for instance, live in Germany and work in Luxembourg would be subject to the treatment, as their income is taxed in Luxembourg. While the dataset from Luxembourg only includes residents, these individuals would be excluded from the data. They could, however, potentially be included in the control group if they took part in the German survey. As there is no available data variable that would allow us to exclude cross-border commuters, we cannot mitigate this problem. However, we do not expect this to have a significant impact on the results, since the share of French and German respondents this applies to is likely to be minimal.

First, we focus on married women of working age. Model 1, which is specified without any additional variables, suggests an increase of 1.34 working hours per week following the reform, relative to comparison countries (p \< 0.1). As with the following models, this effect has to be understood as the intent-to-treat effect, meaning the effect of having the possibility to opt for individual taxation, since, due to data limitations, there is no way to discern those who actually did.

However, when controlling for the level of education and children living in the household in Model 2, the effect largely disappears, the estimate here is negative and statistically insignificant. This suggests that the estimate in the first model partly reflects structural differences in the education levels and household composition of the female working population of Luxembourg as compared to the control countries.

Additionally, we conduct two placebo models to check H2 and confirm whether the theoretical mechanism functions as anticipated. Model 3 is based on married men and does not include control variables. The hypothetical estimate for men is assumed to be closer to zero. As a main primary earner, the effects of substitution and income are expected to offset each other. Contrary to this expectation, the model shows that the reform is associated with a decrease of 2.19 weekly working hours for married men (p \< 0.1). Given that the effect here is even larger in magnitude than in Model 1, there seems to be a shift in working hours around the treatment period that is affecting both men and women.

Lastly, Model 4 focuses on unmarried women. Since they cannot file their taxes jointly, they are not supposed to be affected by the reform. Therefore, any substantial change in working hours would suggest that the shifts are caused by something other than the reform. While the model shows a close-to-zero effect, this result is not statistically significant. Overall, our models cannot demonstrate that the tax reform has increased the working hours of women.

To further substantiate findings from our models, we scrutinize the overall trends of the outcome variable in both Luxembourg and the control countries. The growth of female weekly working hours in those countries is shown in Figure 1. It shows that the working hours of women had already begun to increase substantially in 2013, long before the tax reform. In comparison, the working hours in the control countries increased subtly during the same period. Only Ireland shows a comparable trend. As discussed, this kind of diverging trend violates the parallel trends assumption, which must be upheld in order to derive causal conclusions from the model results. While women’s working hours in Luxembourg briefly continue to increase following the reform, the trend then slows down and is even reversed after 2021. However, our results do not allow us to make credible claims about the causal relationship between the reform and this trend.

## 6 Discussion

Our analysis could not show that the introduction of optional individual taxation for married couples in Luxembourg has a significant impact on the average working hours of women. Contrary to the theoretical prediction and previous studies ([Crossley & Jeon, 2007](#ref-crossley2007); [Kalíšková, 2014](#ref-kaliskova2014); [Selin, 2014](#ref-selin2014)), female labor supply does not appear to be increasing. These results offer several points for discussion:

First, it sheds light on factors endogenous to the tax reform. The tax reform seems not strong enough to change individual behaviors. Rather, the control variables, i.e. the level of education and the presence of children in the household, are expected to play a significant role according to our model results. When these structural factors are translated into policies that facilitate female employment, such as child daycare ([Selin, 2014](#ref-selin2014)), this would offset the opportunity cost of staying in joint taxation. It is therefore worth noting that these confounding factors become more salient in this case.

Second, the findings could be subject to biases related to individual behaviors. Our case builds hypotheses based on the assumption that individuals are rational and utility-maximizing. However, an increase in working hours might not be an optimal choice for some of them. For instance, a decision to get married can be predicated on whether they benefit from the tax reform. The income gap between partners can also be a factor in the decision to work more. In this sense, our model mechanism does not properly represent the sample individuals.

Additionally, some exogenous factors matter. The Covid pandemic, which occurred relatively soon after the reform, changed the economic situation dramatically by freezing labor markets. Therefore, a change in working hours has been contingent upon not only the tax reform, but also some policies to handle the pandemic, which makes it more difficult to predict the pure impact of the treatment.

However, these considerations remain hypothetical, since it is not possible to make any causal claims based on the results of our study. As we can observe in the pre-trends graph, the analysis does not guarantee its findings since it violates the parallel assumption. The fundamental issue is that there is no appropriate country sample, except Ireland, to compare with Luxembourg. From 2012 to 2017, the graph shows a steep rise before the tax reform. This indicates that there were other factors causing an increase in female working hours. Luxembourg introduced two major family policy reforms that might have boosted female working hours: childcare in 2009 and parental leave in 2016. Hence, the effect of these policies should be taken into account along with the tax reform in this period.

Taken together, prospective studies could further explore whether the reform towards optional individual taxation has any causal effect considering the above aspects. This might be achieved by focusing on how many individuals actually take up individual taxation and under what conditions. Another open question is what causes a divergent trend that yields the working hours of women to rise more strongly than in other countries, therefore making Luxembourg an outlier compared to other countries.

## 7 Conclusion

This paper aims to explore the possible effect of the 2018 reform in Luxembourg, which added individual taxation as an option to the existing marital splitting system. We hypothesized that the reform would lead to an increase in female labour supply as measured by a positive change in weekly working hours. To test whether there is a causal relationship between the reform and average female working hours, we applied a difference-in-differences approach to multiple countries, using cross-sectional data from the Luxembourg Income Study.

The findings of our first baseline model showed a statistically significant positive estimate for working hours of the treatment group, married women. When including the control variables on education and children, the effect disappeared. Results from our robustness checks showed an unexpected negative effect on male working hours and a close-to-zero effect for unmarried women, which was closer to our expectations from the theoretical argument.

A possible interpretation is that the optional design of the reform step may create weaker incentives for secondary earners to increase their working hours and generate a higher individual income that could be taxed less, depending on the spouses’ income and the tax brackets they are placed in. Since we did not filter by take-up, the sample also includes couples who continued the marital splitting. This is especially beneficial for couples with unequal earnings and therefore could remain attractive, also since the secondary earner may have specialized less in paid labor over time.

Most problematic when it comes to limitations is the parallel trends assumption not being convincingly fulfilled. Female working hours were already strongly increasing before the reform year in comparison to most countries included. Additionally, other unobserved factors due to the effects of the Covid pandemic might distort our estimate of the tax reform.

Overall, we could not identify convincing evidence that the introduction of optional individual taxation has a significant impact on female working hours. However, the issue remains relevant, as countries such as Germany continue to maintain joint taxation for married couples with the possibility to opt out. Moreover, with individual taxation becoming mandatory for newly married couples in Luxembourg from 2028, future research will have the opportunity to examine whether a more comprehensive shift toward individual taxation leads to significant positive changes in female labor market activity.
